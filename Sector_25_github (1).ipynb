```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the distribution of AQI values
plt.figure(figsize=(10, 6))
sns.histplot(df['AQI'], bins=50, kde=True, color='teal')

plt.title('Distribution of AQI Values', fontsize=14)
plt.xlabel('AQI', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import pandas as pd

# Ensure Timestamp is in datetime format
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# Group AQI by the year extracted from the Timestamp
aqi_yearly_summary = df.groupby(df['Timestamp'].dt.year)['AQI'].describe()
display(aqi_yearly_summary)
```

```python

daily_aqi = df.groupby(df['Timestamp'].dt.date)['AQI'].mean().reset_index()

plt.figure(figsize=(15, 6))
sns.lineplot(data=daily_aqi, x='Timestamp', y='AQI', color='crimson', linewidth=1)

plt.title('Daily Average AQI over Time', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Average AQI', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Extract month names and numbers for correct chronological ordering
df['Month_Num'] = df['Timestamp'].dt.month
df['Month_Name'] = df['Timestamp'].dt.strftime('%B')

# Group by month number and name to compute average AQI, PM2.5, and PM10
monthly_trends = df.groupby(['Month_Num', 'Month_Name'])[['AQI', 'PM2.5 (µg/m³)', 'PM10 (µg/m³)']].mean().reset_index()

# Display the aggregated monthly averages
display(monthly_trends.style.background_gradient(cmap='YlOrRd', subset=['AQI', 'PM2.5 (µg/m³)', 'PM10 (µg/m³)']))
```

```python
# Visualize Month-over-Month AQI Average
plt.figure(figsize=(12, 6))
sns.barplot(data=monthly_trends, x='Month_Name', y='AQI', hue='Month_Name', palette='Oranges_r', legend=False)

plt.title('Average AQI by Month (Seasonal Trend)', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Average AQI', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define raw pollutant parameters and AQI
core_aqi_cols = [
    'PM2.5 (µg/m³)', 'PM10 (µg/m³)', 'NO2 (µg/m³)',
    'NH3 (µg/m³)', 'SO2 (µg/m³)', 'CO (mg/m³)',
    'Ozone (µg/m³)', 'AQI'
]

# Ensure only existing columns in the DataFrame are used
available_cols = [col for col in core_aqi_cols if col in df.columns]

# Calculate correlation matrix
core_corr = df[available_cols].corr()

# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(
    core_corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    square=True,
    linewidths=0.5,
    cbar_kws={'shrink': 0.8}
)

plt.title('Correlation Heatmap: Core Pollutants and AQI', fontsize=14)
plt.tight_layout()
plt.show()
```
